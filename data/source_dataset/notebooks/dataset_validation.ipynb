# Dataset inspection
Load a few rows and inspect coverage and quality flags.

```python
from pathlib import Path
import pandas as pd
root=Path('..')
for name in ['customers','orders','order_items','menu_items','restaurants','ratings','inventory','wastage']:
    df=pd.read_csv(root/'data'/f'{name}.csv', nrows=5)
    print(name, df.shape)
    display(df.head(2))
```

```python
orders=pd.read_csv(root/'data/orders.csv', parse_dates=['Order_Date'])
orders.groupby(orders.Order_Date.dt.day_name()).size().sort_values(ascending=False)
```
